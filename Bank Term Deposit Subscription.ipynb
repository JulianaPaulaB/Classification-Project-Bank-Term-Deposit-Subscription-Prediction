```python
# Reflect your answers here.


# ==========================================
# PART 3: DATA UNDERSTANDING
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# A. METADATA
print("=" * 60)
print("A. METADATA")
print("=" * 60)

print("Dataset Name: Bank Marketing")
print("Data Owner/Author: Sourced from the UCI Machine Learning Repository")
print("Original Research: Moro, Cortez, and Rita (2014)")
print("Source: https://archive.ics.uci.edu/dataset/222/bank+marketing")
print("How Obtained: Downloaded from the official UCI repository")
print("Data Format: CSV (Comma/Semicolon-separated values)")
print("License: Check the dataset's current UCI page for licensing details")
print("Purpose: Predict whether a customer subscribes to a term deposit.")

# Load the dataset
df = pd.read_csv("bank-additional-full.csv", sep=";")

# B. DESCRIBING THE DATA
print("\n" + "=" * 60)
print("B. DESCRIBING THE DATA")
print("=" * 60)

print("\nFirst Five Records:")
display(df.head())

print("\nDataset Dimensions (Rows, Columns):")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nNumerical Summary:")
display(df.describe())

print("\nCategorical Summary:")
display(df.describe(include="object"))

print("\nTarget Variable Distribution:")
print(df["y"].value_counts())

print("\nTarget Variable Percentages:")
print((df["y"].value_counts(normalize=True) * 100).round(2))

# C. VERIFICATION OF DATA QUALITY
print("\n" + "=" * 60)
print("C. VERIFICATION OF DATA QUALITY")
print("=" * 60)

print("\nMissing Values Per Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nNumber of Unique Values Per Column:")
print(df.nunique())

# Check for the dataset's 'unknown' category values
print("\n'Unknown' Values Per Column:")
for column in df.select_dtypes(include="object").columns:
    unknown_count = (
        df[column].astype(str).str.strip().str.lower() == "unknown"
    ).sum()
    if unknown_count > 0:
        print(f"{column}: {unknown_count}")

print("\nTarget Labels:")
print(df["y"].unique())

# D. DATA EXPLORATION
print("\n" + "=" * 60)
print("D. DATA EXPLORATION")
print("=" * 60)

# 1. Distribution of the target variable
df["y"].value_counts().plot(
    kind="bar",
    title="Distribution of Term Deposit Subscriptions",
    xlabel="Subscription Outcome",
    ylabel="Number of Customers",
    rot=0
)
plt.tight_layout()
plt.show()

# 2. Distribution of customer ages
df["age"].plot(
    kind="hist",
    bins=20,
    title="Distribution of Customer Ages",
    xlabel="Age",
    ylabel="Frequency"
)
plt.tight_layout()
plt.show()

# 3. Subscription outcomes by job
subscription_by_job = pd.crosstab(
    df["job"], df["y"], normalize="index"
) * 100

subscription_by_job.plot(
    kind="bar",
    stacked=True,
    title="Subscription Outcomes by Job",
    ylabel="Percentage (%)",
    xlabel="Job",
    figsize=(10, 5)
)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 4. Correlation among numerical features
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(12, 8))
plt.imshow(correlation, cmap="coolwarm", aspect="auto", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Among Numerical Features")
plt.tight_layout()
plt.show()

# 5. Check for unusual numeric values
print("\nPotential Numeric Edge Cases:")
for column in numeric_df.columns:
    print(
        f"{column}: minimum = {df[column].min()}, "
        f"maximum = {df[column].max()}"
    )
```

# Part 3. Perform Data Partitioning



```python

# ==========================================
# PART 3: PERFORM DATA PARTITIONING
# ==========================================

from sklearn.model_selection import train_test_split

# Remove exact duplicate records before partitioning
df_model = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Dataset shape after removing duplicates:", df_model.shape)
print("Duplicate records removed:", df.shape[0] - df_model.shape[0])

# Separate input features (X) and target variable (y)
# Exclude duration because it is known only after the call ends
X = df_model.drop(columns=["y", "duration"])
y = df_model["y"]

print("\nInput features shape:", X.shape)
print("Target variable shape:", y.shape)

# Split into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the partitioning results
print("\nTraining and Testing Results")
print("-" * 40)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nTarget distribution in training set (%):")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTarget distribution in testing set (%):")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

# Part 4. Data Cleaning (when necessary)

```python

# ==========================================
# PART 4: DATA PRE-PROCESSING
# ==========================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Number of numerical features:", len(numeric_features))
print("Numerical features:", numeric_features)

print("\nNumber of categorical features:", len(categorical_features))
print("Categorical features:", categorical_features)

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Fit on training data and transform training and testing data
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("\nPre-processing completed successfully.")
print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
print("Training and testing features are now ready for modeling.")

```

# Part 5. Establishing the Machine Learning Pipeline (Preprocessing and Modelling)

```python
# Dictionary that stores each trained model for the comparison in Part 6
trained_models = {}
```

## Part 5.1. Machine Learning Logistic Regression Model Training and Testing

```python

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.base import clone
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

logistic_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

logistic_model.fit(X_train, y_train)
lr_predictions = logistic_model.predict(X_test)

print("Logistic Regression Results")
print("Accuracy:", round(accuracy_score(y_test, lr_predictions), 4))
print("Precision (Yes):", round(precision_score(y_test, lr_predictions, pos_label="yes"), 4))
print("Recall (Yes):", round(recall_score(y_test, lr_predictions, pos_label="yes"), 4))
print("F1-score (Yes):", round(f1_score(y_test, lr_predictions, pos_label="yes"), 4))

trained_models["Logistic Regression"] = logistic_model
```

## Part 5.2. Machine Learning K-Nearest Neighbors Training and Testing

```python

from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

knn_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)
knn_predictions = knn_model.predict(X_test)

print("K-Nearest Neighbors Results")
print("Accuracy:", round(accuracy_score(y_test, knn_predictions), 4))
print("Precision (Yes):", round(precision_score(y_test, knn_predictions, pos_label="yes"), 4))
print("Recall (Yes):", round(recall_score(y_test, knn_predictions, pos_label="yes"), 4))
print("F1-score (Yes):", round(f1_score(y_test, knn_predictions, pos_label="yes"), 4))

trained_models["K-Nearest Neighbors"] = knn_model
```

## Part 5.3. Machine Learning Decision Tree Classifier Training and Testing

```python

from sklearn.tree import DecisionTreeClassifier

decision_tree_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ))
])

decision_tree_model.fit(X_train, y_train)
dt_predictions = decision_tree_model.predict(X_test)

print("Decision Tree Results")
print("Accuracy:", round(accuracy_score(y_test, dt_predictions), 4))
print("Precision (Yes):", round(precision_score(y_test, dt_predictions, pos_label="yes"), 4))
print("Recall (Yes):", round(recall_score(y_test, dt_predictions, pos_label="yes"), 4))
print("F1-score (Yes):", round(f1_score(y_test, dt_predictions, pos_label="yes"), 4))

trained_models["Decision Tree"] = decision_tree_model

```

## Part 5.4. Machine Learning Random Forest Classifier Training and Testing

```python

from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

random_forest_model.fit(X_train, y_train)
rf_predictions = random_forest_model.predict(X_test)

print("Random Forest Results")
print("Accuracy:", round(accuracy_score(y_test, rf_predictions), 4))
print("Precision (Yes):", round(precision_score(y_test, rf_predictions, pos_label="yes"), 4))
print("Recall (Yes):", round(recall_score(y_test, rf_predictions, pos_label="yes"), 4))
print("F1-score (Yes):", round(f1_score(y_test, rf_predictions, pos_label="yes"), 4))

trained_models["Random Forest"] = random_forest_model


```

# Part 6. Model Comparison

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

model_results = []

for name, model in trained_models.items():
    predictions = model.predict(X_test)

    model_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision (Yes)": precision_score(y_test, predictions, pos_label="yes"),
        "Recall (Yes)": recall_score(y_test, predictions, pos_label="yes"),
        "F1-score (Yes)": f1_score(y_test, predictions, pos_label="yes")
    })

results_df = pd.DataFrame(model_results)
results_df = results_df.sort_values(
    by="F1-score (Yes)",
    ascending=False
).reset_index(drop=True)

display(results_df.round(4))

best_model_name = results_df.loc[0, "Model"]
best_model = trained_models[best_model_name]

print("Best Model:", best_model_name)
print("Selected based on the highest F1-score for the 'yes' class.")
```

# Part 7. Save the Best Model

```python

import joblib
import os

# Identify the best model based on F1-score for the "yes" class
best_model_name = results_df.loc[
    results_df["F1-score (Yes)"].idxmax(), "Model"
]

best_model = trained_models[best_model_name]

# Save the complete model pipeline
model_filename = "best_bank_marketing_model.pkl"
joblib.dump(best_model, model_filename)

print("Best Model:", best_model_name)
print("Model saved successfully as:", model_filename)
print("File location:", os.path.abspath(model_filename))


```
